```python
#analiza ryzyka, dzienny procentowy zwrot
df['Daily_Return'] = df.groupby('Company')['Close'].pct_change()
```

```python
#zmiennosc odch std dziennych zwrotów
volatility = df.groupby('Company')['Daily_Return'].std()
```

```python
#przeskalowanie zmiennosci na roczną mnożymy przez 252 dni handlowych
annualized_volatility = volatility * np.sqrt(252)
```

```python
print("10 NAJBARDZIEJ zmiennych (ryzykownych) spółek")
print(annualized_volatility.sort_values(ascending=False).head(10))
```

```python
print("\n 10 NAJMNIEJ zmiennych (stabilnych) spółek")
print(annualized_volatility.sort_values(ascending=True).head(10))
```

```python
#ZAPIS WYNIKÓW DO NOWEJ RAMKI DANYCH
volatility_df = annualized_volatility.reset_index(name='Volatility')
```

```python
#Wizualizacja
most_volatile_df = volatility_df.sort_values(by='Volatility', ascending=False).head(10)
least_volatile_df = volatility_df.sort_values(by='Volatility', ascending=True).head(10)
```

```python
# Wykres1: Najbardziej zmienne spółki
chart_most_volatile = alt.Chart(most_volatile_df).mark_bar().encode(
    x=alt.X('Company', sort='-y', title='Spółka'), # Sortuj od najwyższej zmienności
    y=alt.Y('Volatility', title='Roczna zmienność (odch. stand.)'),
    tooltip=[
        alt.Tooltip('Company', title='Spółka'),
        alt.Tooltip('Volatility', title='Roczna zmienność', format='.2f')
    ]
).properties(
    title='10 najbardziej zmiennych (ryzykownych) spółek'
).interactive()
```

```python
chart_most_volatile.save('most_volatile_stocks.html')
```

```python
# Wykres 2: Najmniej zmienne spółki
chart_least_volatile = alt.Chart(least_volatile_df).mark_bar().encode(
    x=alt.X('Company', sort='y', title='Spółka'), # Sortuj od najniższej zmienności
    y=alt.Y('Volatility', title='Roczna zmienność (odch. stand.)'),
    tooltip=[
        alt.Tooltip('Company', title='Spółka'),
        alt.Tooltip('Volatility', title='Roczna zmienność', format='.2f')
    ]
).properties(
    title='10 najmniej zmiennych (stabilnych) spółek'
).interactive()
```

```python
chart_least_volatile.save('least_volatile_stocks.html')
```

```python
print("Obliczanie całkowitego zwrotu dla spółek...")
```

```python
df.sort_values(by=['Company', 'Date'], inplace=True)
```

```python
#pierwszą i ostatnią cenę zamknięcia dla KAŻDEJ spółki
first_prices = df.groupby('Company')['Close'].first()
last_prices = df.groupby('Company')['Close'].last()
```

```python
data_counts = df['Company'].value_counts() #zlicza ile dni dla each comp

```

```python
returns_df=pd.DataFrame({'First_Price': first_prices, 'Last_Price': last_prices,
    'Days_Of_Data': data_counts})
```

```python
# USUNIĘCIE spółek które mają mniej niż 750dni
min_days = 750
returns_df_filtered = returns_df[returns_df['Days_Of_Data'] >= min_days].copy()
print(f"Znaleziono {len(returns_df_filtered)} spółek z co najmniej {min_days} dniami danych.")
```

```python
#olbiczenie % zwrotu
returns_df_filtered['Percentage_Return'] = (
    (returns_df_filtered['Last_Price'] - returns_df_filtered['First_Price']) 
    / returns_df_filtered['First_Price']
) * 100
```

```python
print("\n 10 NAJLEPSZYCH inwestycji (Zwycięzcy)")
print(returns_df_filtered.sort_values(by='Percentage_Return', ascending=False).head(10))

print("\n 10 NAJGORSZYCH inwestycji (Przegrani)")
print(returns_df_filtered.sort_values(by='Percentage_Return', ascending=True).head(10))
```

```python
# wizualizacja tego
#przygotowanie danych
top_10_winners = returns_df_filtered.sort_values(by='Percentage_Return', ascending=False).head(10).reset_index()
top_10_losers = returns_df_filtered.sort_values(by='Percentage_Return', ascending=True).head(10).reset_index()
```

```python
#wykres 1 zwyciężcy
chart_winners = alt.Chart(top_10_winners).mark_bar(color='green').encode(
    x=alt.X('Company', sort='-y', title='Spółka'),
    y=alt.Y('Percentage_Return', title='Całkowity zwrot (%)'),
    tooltip=[
        alt.Tooltip('Company', title='Spółka'),
        alt.Tooltip('Percentage_Return', title='Zwrot (%)', format='.1f')
    ]
).properties(
    title='Top 10 Zwycięzców (firmy z >3 latami danych)'
).interactive()
```

```python
# wykres 2 przegrani
chart_losers = alt. Chart(top_10_losers).mark_bar(color='red').encode(
    x=alt.X('Company', sort='y', title='Spółka'),
    y=alt.Y('Percentage_Return', title = 'Całkowity zwrot (%)'),
    tooltip=[
        alt.Tooltip('Company', title='Spółka'),
        alt.Tooltip('Percentage_Return', title='Zwrot (%)', format='.1f')
    ]
).properties(
    title='Top 10 przegranych (firmy z >3latami danych)'
).interactive()
```

```python
chart_winners.save('top_10_winners.html')
chart_losers.save('top_10_losers.html')
```

```python
#przygotowanie danuch do korelacji

```

```python
tickers_to_compare = ['AAPL', 'MSFT', 'GOOGL', 'AMZN', 'NVDA']
```

```python
#Bierzemy nasz dataframe 'df', który zawiera kolumnę 'Daily_Return'
```

```python
if 'Daily_Return' not in df.columns:
    print("Obliczanie 'Daily_Return'...")
    df['Daily_Return'] = df.groupby('Company')['Close'].pct_change()
```

```python
#Filtrujemy dane
df_corr = df[df['Company'].isin(tickers_to_compare)].copy()
```

```python
#Tworzymy tabelę przestawną (pivot)
#    Chcemy, aby daty były indeksem, spółki kolumnami,
#    a wartościami były ich dzienne zwroty ('Daily_Return')

correlation_data = df_corr.pivot_table(
    index='Date', 
    columns='Company', 
    values='Daily_Return'
)
```

```python
correlation_data.head()
```

```python
#Usuwamy pierwszy wiersz, który zawiera NaN (bo nie ma zwrotu dla pierwszego dnia)
correlation_data = correlation_data.dropna()
```

```python
print(correlation_data.head())
```

```python
#heatmap
```

```python
corr_matrix = correlation_data.corr()

print("--- Macierz Korelacji: ---")
print(corr_matrix)
```

```python
corr_matrix.columns.name = None
```

```python
#Teraz "rozprostowujemy" (melt) macierz.
#    'Company' (z indeksu) stanie się kolumną 'Company_1'
#    'variable' (stare nazwy kolumn) stanie się kolumną 'Company_2'
```

```python
corr_matrix_tidy = corr_matrix.reset_index().melt('Company')
corr_matrix_tidy.columns = ['Company_1', 'Company_2', 'Correlation']
```

```python
#zmiana nazw kolumn
corr_matrix_tidy.columns = ['Company_1', 'Company_2', 'Correlation']
```

```python
corr_matrix_tidy.head(10)
```

```python
base = alt.Chart(corr_matrix_tidy).mark_rect().encode(
   
    x=alt.X('Company_1', title='Spółka 1'),
    y=alt.Y('Company_2', title='Spółka 2'),
    
   
    color=alt.Color('Correlation', 
                    scale=alt.Scale(range='heatmap'), # Użyj palety "heatmap"
                    title='Korelacja'
                   ),
    
    # Podpowiedzi
    tooltip=[
        alt.Tooltip('Company_1', title='Spółka 1'),
        alt.Tooltip('Company_2', title='Spółka 2'),
        alt.Tooltip('Correlation', format='.2f')
    ]
).properties(
    title='Mapa Korelacji Dziennych Zwrotów (2018-2023)',
    width=700, height=700
).interactive()
```

```python
text = base.mark_text(baseline='middle').encode(
    text=alt.Text('Correlation', format='.2f'),
    color=alt.value('black')) # Kolor tekstu
```

```python
#Łączymy wykres i tekst
final_heatmap = base + text
```

```python
final_heatmap.save('correlation_heatmap.html')
```
